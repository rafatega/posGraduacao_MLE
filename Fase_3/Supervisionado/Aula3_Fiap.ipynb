# Classificação

```python
# Modelos de Classificação.

import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
df_path = "C:\\Users\\rafaeltegazzini\\Documents\\Pessoais\\posGraduacao_MLE\\Fase_3\\Dados\\iris\\iris.data"
df = pd.read_csv(df_path, header=None)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df
```

```python
# Separando a variável alvo (y) das variáveis de entrada (X)
x = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = df['species']
```

```python
# Dividindo os dados em conjuntos de treino e teste
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=23)

# é importante anotar o random_state para garantir a reprodutibilidade dos resultados, ou seja, para que a divisão dos dados seja a mesma em diferentes execuções do código. O valor 23 é apenas um exemplo e pode ser qualquer número inteiro.
# Os dados de treino (x_train e y_train) serão usados para treinar o modelo, enquanto os dados de teste (x_test e y_test) serão usados para avaliar o desempenho do modelo após o treinamento.
```

## Aplicando os modelos

### Regressão Logística

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
# Fazendo previsões com o modelo treinado
y_pred = model.predict(x_test)
```

```python
# Analizando a 5 primeiras
print("Previsões:", y_pred[:5])
print("Valores Reais:", y_test[:5].values)
```

```python
# Calulando as métricas de avaliação
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"Acurácia: {accuracy:.2f}")
print(f"Precisão: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-Score: {f1:.2f}")
```

```python
# Matriz de confusão
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot()
```

## KNN

```python
from sklearn.neighbors import KNeighborsClassifier
modelo = KNeighborsClassifier(n_neighbors=3)
modelo.fit(x_train, y_train)
```

```python
y_pred_knn = modelo.predict(x_test)
```

```python
# Calculando a acurácia do KNN
accuracy_knn = accuracy_score(y_test, y_pred_knn)
print(f"Acurácia do KNN: {accuracy_knn:.3f}")
```

```python
# Relação de acurácia com a quantidade de vizinhos
number_neighbors = list(range(1, 106))
accuracy_dict = dict()

for n in number_neighbors:
    modelo = KNeighborsClassifier(n_neighbors=n)
    modelo.fit(x_train, y_train)
    y_pred = modelo.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracy_dict[n] = accuracy
```

```python
accuracy_dict
```

```python
plt.plot(number_neighbors, list(accuracy_dict.values()))
plt.xlabel('Número de Vizinhos (k)')
plt.ylabel('Acurácia')
plt.show()
```

## Árvore de Decisão
A Árvore de Decisão é um modelo de aprendizado supervisionado que pode ser usado tanto para classificação quanto para regressão. Ela funciona dividindo os dados em subconjuntos com base em uma série de perguntas, formando uma estrutura em forma de árvore. Cada nó interno da árvore representa uma pergunta sobre um atributo dos dados, e cada folha representa uma classe ou valor de saída.

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
modelo = DecisionTreeClassifier()
modelo.fit(x_train, y_train)
```

```python
y_pred_tree = modelo.predict(x_test)
```

```python
accuracy_tree = accuracy_score(y_test, y_pred_tree)
print(f"Acurácia da Árvore de Decisão: {accuracy_tree:.3f}")
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12, 8))
plot_tree(modelo,
          filled=True, 
          feature_names=x.columns, 
          class_names=modelo.classes_)
```

## Support Vector Machine (SVM)

```python
from sklearn.svm import SVC
modelo_svc = SVC()
modelo_svc.fit(x_train, y_train)
y_pred_svc = modelo_svc.predict(x_test)
accuracy_svc = accuracy_score(y_test, y_pred_svc)
print(f"Acurácia do SVC: {accuracy_svc:.3f}")
```

## Redes Neurais Artificiais
As redes neurais artificiais são modelos de aprendizado inspirados no funcionamento do cérebro humano. Elas consistem em camadas de neurônios artificiais que processam os dados de entrada e aprendem a realizar tarefas específicas, como classificação ou regressão. As redes neurais são especialmente eficazes em lidar com dados complexos e não lineares, tornando-as uma escolha popular para uma ampla variedade de problemas de aprendizado de máquina.

### Perceptron
O perceptron é um modelo de classificação linear que pode ser usado para resolver problemas de classificação binária. Ele funciona ajustando um hiperplano que separa as classes no espaço de características. O perceptron é um modelo simples, mas pode ser eficaz para problemas lineares. No entanto, ele não é capaz de lidar com problemas não lineares, o que limita sua aplicabilidade em muitos casos do mundo real.

```python
from sklearn.linear_model import Perceptron
modelo_perceptron = Perceptron()
modelo_perceptron.fit(x_train, y_train)
y_pred_perceptron = modelo_perceptron.predict(x_test)
accuracy_perceptron = accuracy_score(y_test, y_pred_perceptron)
print(f"Acurácia do Perceptron: {accuracy_perceptron:.3f}")
```

### Multilayer Perceptron (MLP)
O MLP é uma extensão do perceptron que inclui múltiplas camadas de neurônios, permitindo que ele aprenda representações mais complexas dos dados. O MLP é capaz de lidar com problemas não lineares e é amplamente utilizado em tarefas de classificação e regressão. Ele é treinado usando o algoritmo de retropropagação, que ajusta os pesos das conexões entre os neurônios para minimizar o erro de previsão.

```python
from sklearn.neural_network import MLPClassifier
modelo_mlp = MLPClassifier(max_iter=1000)
modelo_mlp.fit(x_train, y_train)
y_pred_mlp = modelo_mlp.predict(x_test)
accuracy_mlp = accuracy_score(y_test, y_pred_mlp)
print(f"Acurácia do MLP: {accuracy_mlp:.3f}")
```
